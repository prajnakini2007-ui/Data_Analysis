```python
num_cols = df.select_dtypes(include=['int64', 'float64'])

for col in num_cols.columns:
    print(f"\nColumn: {col}")
    print("Mean   :", num_cols[col].mean())
    print("Median :", num_cols[col].median())
    print("Mode   :", num_cols[col].mode()[0])
```

```python
num_cols = df.select_dtypes(include=['int64', 'float64'])

for col in num_cols.columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    print(f"\nColumn: {col}")
    print(f"Number of Outliers: {len(outliers)}")
    
    if len(outliers) > 0:
        print(outliers[[col]])
```

```python
import seaborn as sns
columns = [
    'Customer_Age',
    'Number_of_Travellers',
    'Number_of_Adults',
    'Number_of_Children',
    'Hotel_Rating',
    'Customer_Rating'
]

for col in columns:
    plt.figure(figsize=(6,4))
    plt.hist(df[col], bins=10, color='skyblue', edgecolor='black')
    plt.title(f'Frequency Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()
```

```python
rating_counts = df['Number_of_Travellers'].value_counts().sort_index()

plt.figure(figsize=(8,5))
plt.bar(rating_counts.index, rating_counts.values, color='pink', edgecolor='black')

plt.title('Customer Rating Distribution')
plt.xlabel('Customer Rating')
plt.ylabel('Number of Customers')
plt.xticks(rating_counts.index)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

```python
df['Booking_Date'] = pd.to_datetime(df['Booking_Date'])

df['Month'] = df['Booking_Date'].dt.month_name()

monthly_bookings = df['Month'].value_counts().reindex([
    'January', 'February', 'March', 'April', 'May', 'June',
    'July', 'August', 'September', 'October', 'November', 'December'
])

plt.figure(figsize=(10,5))
monthly_bookings.plot(kind='bar', color='yellow')

plt.title('Monthly Booking Trends')
plt.xlabel('Month')
plt.ylabel('Number of Bookings')
plt.xticks(rotation=45)
plt.show()
```

```python
# Top destinations
destination_counts = df['Destination_Country'].value_counts()

print(destination_counts)

plt.figure(figsize=(10,5))
destination_counts.plot(kind='bar', color='skyblue')
plt.title('Popular Destinations')
plt.xlabel('Destination')
plt.ylabel('Number of Bookings')
plt.xticks(rotation=45)
plt.show()
```

```python
hotel_counts = df['Hotel_Name'].value_counts()

print(hotel_counts)
```

```python
transport_counts = df['Transportation_Type'].value_counts()

print(transport_counts)

plt.figure(figsize=(8,5))
transport_counts.plot(kind='bar', color='orange')
plt.title('Popular Transportation Types')
plt.xlabel('Transportation Type')
plt.ylabel('Number of Bookings')
plt.xticks(rotation=45)
plt.show()
```

```python
print("Most Popular Destination:",
      df['Destination_Country'].value_counts().idxmax())

print("Most Popular Accommodation Type:",
      df['Hotel_Name'].value_counts().idxmax())

print("Most popular Transportation type:",
      df['Transportation_Type'].value_counts().idxmax())
```

```python
corr_data = df[['Total_Trip_Cost', 'Number_of_Nights', 'Discount_Amount', 'Customer_Rating']]

# Calculate correlation matrix
correlation_matrix = corr_data.corr()

# Display correlation values
print(correlation_matrix)

# Create heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)
```
